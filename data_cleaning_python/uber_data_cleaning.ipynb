# Uber Customer Reviews: Data Cleaning Project

## Introduction
This project demonstrates data cleaning techniques applied to a messy dataset of Uber customer reviews.
The dataset includes missing values, comments in different languages, and other potential inconsistencies.

```python
import pandas as pd
import numpy as np

# Loading the dataset
data_path = "https://raw.githubusercontent.com/orhunsendur/Portfolio/main/data_cleaning_python/uber_reviews.csv"
df = pd.read_csv(data_path)
```

## Data Cleaning Process

```python
# Remove unnecessary columns and handle null values
columns_to_drop = ['userImage', 'replyContent', 'repliedAt']
df_clean = df.drop(columns=columns_to_drop)

# Convert timestamp
# Ensure the 'at' column is in datetime format
df_clean['at'] = pd.to_datetime(df_clean['at'])

# Clean text content
# Strip whitespace from the 'content' column
df_clean['content'] = df_clean['content'].str.strip()

# Fill missing version numbers
# Replace NaN values in version columns with 'N/A'
df_clean['reviewCreatedVersion'] = df_clean['reviewCreatedVersion'].fillna('N/A')
df_clean['appVersion'] = df_clean['appVersion'].fillna('N/A')

# Basic statistics after cleaning
print("Dataset shape after initial cleaning:", df_clean.shape)
print("\nSample of cleaned data:")
print(df_clean.head())

# Check value distributions
# Display score distribution
print("\nScore distribution:")
print(df_clean['score'].value_counts().sort_index())

# Final dataset overview
# Save the cleaned dataset for further analysis
df_clean.to_csv("clean_uber_reviews.csv", index=False)
```

## Conclusion
- Removed unnecessary columns and handled missing values.
- Converted timestamps to a uniform date-time format.
- Cleaned text data and filled missing version numbers.
- Dataset is now ready for further analysis or visualization.
